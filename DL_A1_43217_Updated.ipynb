### Features and functionality of TensorFlow

- **Open-source framework:** Used to develop and train machine learning and deep learning models.
- **Tensor operations:** Represents data using multidimensional arrays called tensors and supports mathematical operations on them.
- **Automatic differentiation:** Computes gradients needed to train models.
- **Hardware support:** Can use CPUs and compatible GPUs for computation.
- **Model development:** Provides APIs and tools for building, training, evaluating, and deploying models.

The code below demonstrates basic tensor creation and element-wise operations.


```python
# Demonstrating TensorFlow tensor operations
a = tf.constant([1, 2, 3])
b = tf.constant([4, 5, 6])

print("Tensor A:", a.numpy())
print("Tensor B:", b.numpy())
print("Addition:", (a + b).numpy())
print("Multiplication:", (a * b).numpy())
```

### Features and functionality of Keras

- **High-level API:** Makes neural-network development simpler and more readable.
- **Model-building APIs:** Supports the Sequential API and the Functional API.
- **Layer support:** Includes common layers such as Dense, Convolutional, Pooling, and Dropout layers.
- **Training workflow:** The `compile()`, `fit()`, and `evaluate()` methods configure, train, and evaluate models.
- **Backend integration:** In this notebook, Keras is used with the TensorFlow-based environment.

The MNIST dataset above contains 60,000 training images and 10,000 testing images, each 28 × 28 pixels. The following example defines a small classifier and displays its architecture; it does not train the model.


```python
from keras.models import Sequential
from keras.layers import Input, Flatten, Dense

# Define a simple neural network for 28 x 28 MNIST images
model = Sequential([
    Input(shape=(28, 28)),
    Flatten(),
    Dense(128, activation="relu"),
    Dense(10, activation="softmax")
])

model.compile(
    optimizer="sgd",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
import pytensor
import pytensor.tensor as T

# Define symbolic variables
a = T.dscalar("a")
b = T.dscalar("b")

# Define a symbolic addition operation
c = a + b

# Compile the symbolic expression into a function
add_function = pytensor.function([a, b], c)

# Evaluate the function
result = add_function(5.0, 7.0)

print("Result of symbolic addition:", result)
```

## Features and functionality of PyTensor

- **Symbolic computation:** Mathematical expressions can be defined symbolically before they are evaluated.
- **Function compilation:** Expressions can be compiled into callable functions, as demonstrated in the addition example above.
- **Automatic differentiation:** Supports gradient calculations useful in optimization and machine learning.
- **Array computation:** Supports mathematical operations on multidimensional arrays and tensors.
- **Successor to Theano:** PyTensor is a modern successor to Theano that continues its symbolic computation approach and is suitable for mathematical computation in Python.

In the existing example, `T.dscalar()` declares symbolic scalar inputs, `a + b` builds a symbolic expression, and `pytensor.function([a, b], c)` compiles it into a function that returns `12.0` for inputs `5.0` and `7.0`.

### Features and functionality of PyTorch

- **Tensor computation:** Provides multidimensional tensors and mathematical operations.
- **Automatic differentiation:** Its `autograd` system calculates gradients for model training.
- **Neural-network tools:** The `torch.nn` module provides layers, loss functions, and other building blocks.
- **Flexible training:** Supports custom training loops and dynamic model development.
- **Hardware support:** Can use compatible CUDA-enabled GPUs; `False` from `torch.cuda.is_available()` means CUDA was not available in the environment where that cell was run. CPU-based computation is still possible.
